Facemask wearing covid model

Dataset with state and covid 7 day rolling cases and deaths are considered here for the analysis.

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

yougov = pd.read_csv("yougov_oxcgrt_rolling_cleaned_preprocessed_withoutstate.csv") # dataset without dummy variables for state 

print(yougov.columns)
print(f"\nyougov columns:",len(yougov.columns))
```

Data Splitting

Face mask wearing

```python
# Data splitting 80-20%  - face mask 

from sklearn.model_selection import train_test_split

# 80% Train  20% Test 
state_train_facemask, state_test_facemask = train_test_split(
    yougov,test_size=0.20,random_state=42,stratify=yougov['face_mask_binary']
)


print("facemask:")
print("Training:", len(state_train_facemask))
print("Test:", len(state_test_facemask))


state_train_facemask.to_csv("state_train_facemask.csv", index=False)
state_test_facemask.to_csv("state_test_facemask.csv", index=False)
```

General protective behaviour

```python
# Data splitting 80-20%  - face mask 

from sklearn.model_selection import train_test_split

# 80% Train  20% Test 
state_train_general_behav, state_test_general_behav = train_test_split(
    yougov,test_size=0.20,random_state=42,stratify=yougov['general_protective_behavior_binary']
)

print("\ngeneral protective behavior binary:")
print("Training:", len(state_train_general_behav))
print("Test:", len(state_test_general_behav))


state_train_general_behav.to_csv("state_train_general_behav.csv", index=False)
state_test_general_behav.to_csv("state_test_general_behav.csv", index=False)
```

Target class Imbalance

facemask

```python
for state in sorted(state_train_facemask["state"].unique()):

    train = state_train_facemask[state_train_facemask["state"] == state]

    print("\n", state)
    print(train["face_mask_binary"].value_counts())
```

general protective behaviour

```python
for state in sorted(state_train_general_behav["state"].unique()):

    train = state_train_general_behav[state_train_general_behav["state"] == state]

    print("\n", state)
    print(train["general_protective_behavior_binary"].value_counts())
```

```python
print('\nAll sets:')
print(f"\nface_mask_binary:",yougov['face_mask_binary'].value_counts(normalize=True)*100)

print(f"\ngeneral_protective_behavior_binary:",yougov['general_protective_behavior_binary'].value_counts(normalize=True)*100)



print('\nTrain sets:')

print(f"\nface_mask_binary:",state_train_facemask['face_mask_binary'].value_counts(normalize=True)*100)

print(f"\ngeneral_protective_behavior_binary:",state_train_general_behav['general_protective_behavior_binary']
      .value_counts(normalize=True)*100)
```
