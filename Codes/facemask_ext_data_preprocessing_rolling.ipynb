```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
yougov = pd.read_csv("yougov_oxcgrt_rolling_merged.csv") 

print(yougov.columns)
print(f"\nyougov columns:",len(yougov.columns))
```

```python
# remove cases_daily and deaths_daily columns since 7day rolling averages are considered.
yougov=yougov.drop(columns=["cases_daily","deaths_daily"])
```

Rename the columns for readability

```python
col_names = {
    'i2_health' : 'Non-household contacts',
    'r1_1': 'Perceived Severity',
    'r1_2': 'Perceived Susceptibility',
    'i9_health' : 'Isolate if unwell',
    'i11_health': 'Isolate if instructed', 
    'WCRex2':"Confidence in Goverment's response",
    'cantril_ladder': 'Wellbeing',
    'PHQ4_1' : 'Pleasure in doing things',
    'PHQ4_2':'Feeling down/depressed',
    'PHQ4_3': 'Feeling nervous/anxious',
    'PHQ4_4':"Can't control worrying",
    'WCRex1' : 'How Government handle situation'
}

yougov=yougov.rename(columns=col_names)

print(yougov.columns)
print(f"\nyougov columns:",len(yougov.columns))
```

Columns have been successfully renamed.

```python
yougov.to_csv("yougov_oxcgrt_rolling_merged_renamed.csv")
```
